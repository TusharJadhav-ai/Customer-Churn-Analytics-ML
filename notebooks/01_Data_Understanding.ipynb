## 12. Target Variable Analysis

The target variable is the outcome that the machine learning model will predict. In this project, the target variable is **Churn**, which indicates whether a customer has discontinued the company's services.

Understanding the distribution of the target variable helps identify potential class imbalance and provides an initial view of customer retention performance.

```python
# Count of each target class
df["Churn"].value_counts()
```

```python
# Percentage distribution
df["Churn"].value_counts(normalize=True).mul(100).round(2)
```

```python
churn_summary = pd.DataFrame({
    "Count": df["Churn"].value_counts(),
    "Percentage": (df["Churn"].value_counts(normalize=True) * 100).round(2)
})

churn_summary
```

## 📌 Business Observation

**Observation**

Approximately **73%** of customers remained with the company, while **27%** discontinued the service. This indicates that the dataset contains more retained customers than churned customers.

**Business Impact**

The target variable exhibits a moderate class imbalance. While this reflects real business conditions, it should be considered during model evaluation to ensure the predictive model performs well for both retained and churned customers.

## 13. Variable Layout (SPSS Style)

This section provides a structured overview of each variable in the dataset, similar to the **Variable View** in SPSS. It summarizes the data type, number of unique values, business description, and analytical role of each variable.

Creating a variable layout improves dataset documentation, supports reproducibility, and provides a quick reference for future analysis and machine learning.

```python
# Business descriptions for each variable
variable_description = {
    "customerID": "Unique customer identifier",
    "gender": "Customer gender",
    "SeniorCitizen": "Senior citizen status (0 = No, 1 = Yes)",
    "Partner": "Whether the customer has a partner",
    "Dependents": "Whether the customer has dependents",
    "tenure": "Number of months the customer has stayed",
    "PhoneService": "Phone service subscription",
    "MultipleLines": "Multiple phone lines subscription",
    "InternetService": "Internet service type",
    "OnlineSecurity": "Online security service",
    "OnlineBackup": "Online backup service",
    "DeviceProtection": "Device protection plan",
    "TechSupport": "Technical support service",
    "StreamingTV": "Streaming TV subscription",
    "StreamingMovies": "Streaming movie subscription",
    "Contract": "Customer contract type",
    "PaperlessBilling": "Paperless billing option",
    "PaymentMethod": "Payment method used",
    "MonthlyCharges": "Monthly subscription charges",
    "TotalCharges": "Total amount charged",
    "Churn": "Customer churn status"
}

# Define analytical role
variable_role = {
    "customerID": "Identifier",
    "Churn": "Target"
}

variable_layout = pd.DataFrame({
    "Variable": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Unique Values": df.nunique().values,
    "Description": [variable_description[col] for col in df.columns],
    "Role": [variable_role.get(col, "Predictor") for col in df.columns]
})

variable_layout
```
